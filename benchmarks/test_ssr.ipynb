```python
# method_AG = ['Google Sycamore']
rows = 5
cols = 4
method_AG = [f"Grid {rows}*{cols}"]
# method_AG = ['IBM Heron']
# method_AG = ['IBM Rochester']

parent_directory = str(Path.cwd().parent)
input_folder = "".join([parent_directory, "/ucc/ssr/Data/RevLib/Sabre/"])

output_xlsx = f"/Users/mistywahl/Documents/GitHub/ucc/ucc/ssr/Results/ucc_test/{str(method_AG).replace(' ', '')}.xlsx"

output_qasm_folder = "".join([parent_directory, "/ucc/ssr/Results/ucc_test/qasm/"])
```

```python
process_quantum_circuits(
    input_folder, method_AG, output_xlsx, output_qasm_folder
)
```

```python
def check_ssr_logical_equivalence(input_circuit, output_circuit):
    sv1 = Statevector(input_circuit)
    sv2 = Statevector(output_circuit)
    return sv1.equiv(sv2, atol=0.01)


input_circuits = [qasm2.load("".join([input_folder, f]), custom_instructions=qasm2.LEGACY_CUSTOM_INSTRUCTIONS) 
                      for f in os.listdir(input_folder)
                 ]
output_circuits = [qasm2.load("".join([output_qasm_folder, f]), custom_instructions=qasm2.LEGACY_CUSTOM_INSTRUCTIONS) 
                       for f in os.listdir(output_qasm_folder)
                  ]
result = [check_ssr_logical_equivalence(ic, oc) for ic, oc in zip(input_circuits, output_circuits)]
result
```
